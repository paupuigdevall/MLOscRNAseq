```python
outdir="saved/scanpy/"
Branch_adata = sc.read_h5ad(outdir+"mlo_with_new_annot_wPaga_foetal.h5ad")
```

```python
BranchBCs = Branch_adata.obs_names[Branch_adata.obs["endpoint_Astrocytes"] == 1].tolist()
Branchtag = "Astro"
```

```python
##PCW10
#Samples from "F10121620_C" and "F10121620_D" are already merged
```

```python
adata=sc.read_h5ad(outdir+"mlo_with_new_annot_rawdata_foetal.h5ad")
print(adata.shape)
mask_pcw10 = adata.obs["donorDimensions"]=="PCW10"
adata_pcw10 = adata[mask_pcw10, :].copy()
print(adata_pcw10.shape)

adata_pcw10.var_names_make_unique()

#Subset by cells present only in the astrocyte trajectory
adata_pcw10 = adata_pcw10[pd.Series(list(set(adata_pcw10.obs_names).intersection(set(BranchBCs))))]
print(adata_pcw10.shape)
sc.pp.filter_genes(adata_pcw10, min_cells=3)

sc.pp.normalize_total(adata_pcw10, target_sum=2e4, layers = "all")
sc.pp.log1p(adata_pcw10)
sc.pp.highly_variable_genes(adata_pcw10, n_top_genes=2000, batch_key = "midBrainId")

HVGpcw10 = adata_pcw10.var[adata_pcw10.var.highly_variable_nbatches >= 2].index.tolist()
print(len(HVGpcw10))

del adata
```
